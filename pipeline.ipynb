#Limpeza e Tratamento das Tabelas

##Importação das tabelas

```python
import pandas as pd
df_games = pd.read_csv('games.csv')
df_platforms = pd.read_csv('platform_summary.csv')
df_publisher = pd.read_csv('publisher_summary.csv')
df_genres = pd.read_csv('genre_summary.csv')
df_platforms = pd.read_csv('platform_summary.csv')
```

##Tratamento da tabela Platform

```python
colunas_sales = ['game_id','na_sales_million','eu_sales_million','jp_sales_million','other_sales_million','global_sales_million','estimated_revenue_million_usd']
colunas_reviews = ['metacritic_score','critic_review_count','user_score','user_review_count']
colunas_publisher = ['publisher_region','publisher_tier']
```

```python
df_platforms.reset_index(inplace=True)
df_platforms.index += 1
df_sales = pd.DataFrame()
df_sales= df_games[colunas_sales].copy()
```

```python
df_reviews = pd.DataFrame(data = df_games[colunas_reviews].copy(),index=range(1,len(df_games)+1))
```

```python
df_games['platform'] = df_games['platform'].map(df_platforms.set_index('platform')['index'])
df_games = df_games.drop(columns=['platform_type','platform_maker','platform_generation'])
colunas_sales.remove('game_id')
```

```python
df_publisher.reset_index(inplace=True)
df_games['publisher'] = df_games['publisher'].map(df_publisher.set_index('publisher')['index'])
```

```python
df_games = df_games.drop(columns=colunas_publisher)
df_games = df_games.drop( columns = colunas_reviews)
df_games = df_games.drop(columns= colunas_sales)
```

```python
df_genres.reset_index(inplace=True)
```

```python
df_games['genre'] = df_games['genre'].map(df_genres.set_index('genre')['index'])
```

```python
df_games.rename(columns={'platform':'platform_id','publisher':'publisher_id','genre':'genre_id'}, inplace=True)
```

```python
for col in df_games.columns[9:17]:
    df_games[col] = df_games[col].map({1: True, 0: False})
```
